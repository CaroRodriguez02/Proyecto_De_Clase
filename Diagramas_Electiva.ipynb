#**DIAGRAMAS NUMPY**

```python
import numpy as np
import matplotlib.pyplot as plt

#Ejercicio 1: crear valores
valores = np.linspace(-4, 4, 15)
print(valores)

#Ejercicio 2: función cúbica
x = np.linspace(-4, 4, 25)
y = x**3
plt.plot(x, y)
plt.show()

# Ejercicio 3: función cúbica con formato
plt.plot(x, y, color="green", marker="o", linestyle=":")
plt.title("$f(x)=x^3$")
plt.xlabel("$x$")
plt.ylabel("$x^3$")
plt.grid()
plt.show()

# Ejercicio 4: gráfico de dispersión
plt.scatter(x, y)
plt.title("Gráfica de dispersión")
plt.xlabel("$x$")
plt.ylabel("$x^3$")
plt.show()

# Ejercicio 5: gráfico circular
datos = np.abs(y[:6])
plt.pie(datos)
plt.title("Distribución de datos")
plt.show()

# Ejercicio 6: gráfico de caja
plt.boxplot(y)
plt.title("Diagrama de caja")
plt.ylabel("Valores")
plt.show()

# Ejercicio 7: función seno
x = np.linspace(0, 12, 150)
seno = np.sin(x)
plt.plot(x, seno)
plt.title("$F(x)=Sen(x)$")
plt.xlabel("$x$")
plt.ylabel("$Sen(x)$")
plt.xlim(0, 12)
plt.ylim(-1.5, 1.5)
plt.grid()
plt.show()

# Ejercicio 8: seno y coseno
x = np.linspace(0, 12, 150)
seno = np.sin(x)
coseno = np.cos(x)
plt.plot(x, seno, label="$Sen(x)$")
plt.plot(x, coseno, label="$Cos(x)$")
plt.title("Funciones seno y coseno")
plt.xlabel("$x$")
plt.ylabel("Valor")
plt.legend()
plt.grid()
plt.show()

# Ejercicio 9: subplots
plt.subplot(2, 1, 1)
plt.plot(x, seno, label="$Sen(x)$")
plt.title("Función Seno")
plt.xlabel("$x$")
plt.legend()
plt.grid()
plt.subplot(2, 1, 2)
plt.plot(x, coseno, color="orange", label="$Cos(x)$")
plt.title("Función Coseno")
plt.xlabel("$x$")
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```
